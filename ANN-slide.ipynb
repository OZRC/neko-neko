### <font color='grey'>Modeling complex problems with ANN</font>
The basic concept behind artificial NNs is built upon **hypotheses and models of how the human brain works** to solve complex tasks. Although artificial NNs have only become popular in recent years, their earliest conceptual foundations date back to the early 1940s, when Warren McCulloch and Walter Pitts first described how neurons operate. *{A logical calculus of the ideas immanent in nervous activity, W. S. McCulloch and W. Pitts. The Bulletin of Mathematical Biophysics, 5(4):115–133, 1943.}*

However, after several decades following the first experimental artificial neural model *(Rosenblatt’s perceptron in the 1950s)*, many researchers and graduate students gradually became discouraged and lost interest in NNs, because at that time no one was able to produce a good solution for multi-layer neural networks.

*For readers interested in the history of Artificial Intelligence (AI), machine learning, or neural networks in particular, I encourage you to read the very detailed English article on Wikipedia about these periods, titled [AI winters](https://en.wikipedia.org/wiki/AI_winter).*

Finally, in 1986, when D. E. Rumelhart, G. E. Hinton, and R. J. Williams published the backpropagation algorithm for training neural networks more effectively—which we will discuss in more detail later in this chapter—neural networks were brought back into the race in artificial intelligence research. Along with the rapid development of hardware (GPU, TPU), neural networks have become more popular than ever. This led to the emergence of the concept of architectures and algorithms known as **deep learning**, which we often hear about today. It is a hot topic not only in academic research but also one that has received substantial investment from major technology companies such as Facebook, Microsoft, Amazon, Uber, and Google.

To this day, complex neural networks supported by deep learning algorithms are considered state-of-the-art (SOTA) solutions for solving challenging problems such as image recognition, speech recognition, and natural language processing. Common examples of everyday products powered by deep neural networks include Google Image Search and Google Translate— a smartphone application that can automatically recognize text in images and translate it in real time into more than 20 languages.

### <font color='grey'>Recap: Single-layer neural network</font>
Before going into more detail about multi-layer neural network architectures, let us first review some concepts of single-layer neural networks that were introduced in Chapter 2, *Training Simple Machine Learning Algorithms for Classification*, known as the **ADAptive LInear NEuron (Adaline)** algorithm.

![figure 1](images/ANN/SLNN.png)
<div align="center">Figure 1. Architecture of a single-layer artificial neural network</div>  

In Chapter 2, we developed the Adaline algorithm to perform binary classification and used the gradient descent optimization algorithm to update the model’s weight coefficients using the training set. In each epoch, we update the weight vector $w$ using the following update rule:
$$
\begin{aligned}
 w:&=w+\Delta{w}, \text{ where $\Delta{w} = -\eta\nabla{J(w)}$}
\end{aligned}
$$
In other words, we computed the gradient based on the entire training set and updated the model’s weights by taking a step in the opposite direction of the gradient (which can be interpreted as the slope) of $\nabla{J(w)}$. To optimize the model weights, we optimized a predefined objective function; here, we used the Sum of Squared Errors (SSE) as the cost function $J(w)$. Next, we also defined a learning rate to balance the learning speed and avoid the risk of overshooting the global minimum of the cost function.

In gradient descent optimization, we update all weights simultaneously after each epoch, and we also derived the partial derivative of each weight $w_j$ using the following formula:
$$
\begin{aligned}
 \frac{\partial}{\partial{w_j}}J(w)&=-\sum_i(y^{(i)}-a^{(i)})x^{(i)}_j
\end{aligned}
$$
where $y^{(i)}$ is the target class label of sample $x^{(i)}$, and $a^{(i)}$ is the activation of the neuron.

Furthermore, we defined the activation function $\phi(.)$ as follows:
$$
\begin{aligned}
 \phi(z) = z = a
\end{aligned}
$$
where the input $z$ is a weighted linear combination of the connections between the input layer and the output layer:
$$
\begin{aligned}
 z = \sum_j w_j x_j = W^TX
\end{aligned}
$$
While we used the activation $\phi(z)$ to compute the gradient updates, we also defined a threshold to convert the continuous output values into binary values for the binary classification task:
$$
\begin{aligned}
 y = \begin{cases}
        1 \: \text{if $g(z)\geq 0$}\\
        -1 \: \text{otherwise}
     \end{cases}
\end{aligned}
$$
At the same time, we also learned a trick to speed up the learning process of the model, called **stochastic gradient descent (SGD)** optimization. SGD approximates the cost using a single training sample (online learning) or a small subset drawn from the training samples (mini-batch learning). In addition to helping the model learn faster, this is due to the more frequent weight updates compared to standard gradient descent (GD).

### <font color='grey'>Introduction to multi-layer neural network architectures</font>
In this section, we will learn how to connect multiple individual neurons into a multi-layer neural network; this special type of network is called a Multi-Layer Perceptron (MLP). The figure below illustrates the concept of an MLP with three layers:

![](images/ANN/FF.png)
<div align="center">Figure 2. Architecture of a multi-layer artificial neural network</div>  

The MLP shown above consists of an input layer, a hidden layer, and an output layer. The units in the hidden layer are fully connected to the input layer, and similarly, the output layer is fully connected to the preceding hidden layer. If such a network has more than one hidden layer, we call it a **deep artificial neural network**.

As shown in the previous figure, we denote the activation of the $i$-th unit in the $l$-th layer by $a^{(l)}_i$. To make mathematical formulation and experimental coding easier, we will not use numerical indices to refer to layers or unit positions within a layer. Instead, we define superscripts such as *in* for the input layer, *h* for hidden layers, and *out* for the output layer. Note that, for coding convenience, the activation units $a^{(in)}_0$ and $a^{(h)}_0$ are bias units, which are set by default to 1:
$$
\begin{aligned}
f&=  \begin{bmatrix}
        a^{(in)}_0\\
        a^{(in)}_1\\
        \vdots\\
        a^{(in)}_m 
    \end{bmatrix} &= \begin{bmatrix}
        1\\
        x^{(in)}_1\\
        \vdots\\
        x^{(in)}_m 
    \end{bmatrix}
\end{aligned}
$$

Each unit in layer $l$ is connected to all units in layer $l+1$ by a set of weights. For example, the connection between the $k$-th unit in layer $l$ and the $j$-th unit in layer $l+1$ is denoted by $w^{(l)}_{k,j}$. Referring back to the previous figure, we denote the weight matrix connecting the input layer and the hidden layer as $W^{(h)}$, and the weight matrix connecting the hidden layer and the output layer as $W^{(out)}$.

Looking back at the single-layer neural model, we see that having one unit in the output layer is suitable for binary classification problems. In Figure 2, we are shown a more general neural network structure that allows us to perform multi-class classification by generalizing the One-vs-All (OvA) technique. To better understand how this works, let us recall the one-hot representation introduced in Chapter 4, *Building Good Training Datasets – Data Preprocessing*.  
For example, if we have three labels defined as $[0,1,2]$, then after one-hot encoding we obtain the following representations:
$$
\begin{aligned}
0&=  \begin{bmatrix}
        1\\
        0\\
        0 
    \end{bmatrix}, 1 =\begin{bmatrix}
        0\\
        1\\
        0 
    \end{bmatrix}, 2 =\begin{bmatrix}
        0\\
        0\\
        1
    \end{bmatrix}
\end{aligned}
$$

Okay, so with multi-layer neural networks and one-hot encoding, we can define multi-class classification problems under the assumption that each data point belongs to exactly one class.

Now let us summarize the concepts introduced in this section with a visual illustration of a 3-4-3 MLP:

![figure 3](images/ANN/FF2.png)

After understanding the basic concepts of MLPs, we can now explore how an MLP learns. Similar to a single-layer neural network, we can summarize the learning process in three steps:
- Starting from the input layer, we perform forward propagation (also called feed-forward) of the training data patterns through the current MLP architecture to compute the outputs.
- Based on the network’s output, we compute the error that needs to be minimized using a cost function (sometimes referred to as a loss function).
- We perform backpropagation of the error, based on which we compute the partial derivatives of each weight in the network and update the model.

### <font color='grey'>Activating a neural network via forward propagation (feed-forward)</font>
In this section, we will study the feed-forward process used to compute the output of an MLP model.

Let us now go through each step to better understand the feed-forward process for computing the output. Since each unit in the hidden layer is fully connected to all units in the previous layer, we compute the activation of the hidden unit $a^{(h)}_1$ as follows:
$$
\begin{aligned}
z^{(h)}_1 &= a^{(in)}_0 w^{(h)}_{0,1} + a^{(in)}_1 w^{(h)}_{1,1}+...+a^{(in)}_m w^{(h)}_{m,1} \\
a^{(h)}_1 &= \phi(z^{(h)}_1)
\end{aligned}
$$
where $z^{(h)}_1$ is the input to the unit and $\phi(.)$ is an activation function. To handle complex tasks such as image classification or speech processing, we use nonlinear functions as activation functions. For example, the sigmoid function introduced in Chapter 3, *A Tour of Machine Learning Classifiers Using scikit-learn*, is given by:
$$
\begin{aligned}
\phi(z) = \frac{1}{1+e^{-z}}
\end{aligned}
$$

For convenience, efficiency in coding, and better readability, we rewrite the activation function in a more compact form by leveraging concepts from linear algebra. This allows us to vectorize computations using NumPy instead of writing tens or billions of operations with Python loops:
$$
\begin{aligned}
\boldsymbol{z}^{(h)} &= \boldsymbol{a}^{(in)}W^{(h)}\\
\boldsymbol{a}^{(h)} &= \phi(\boldsymbol{z}^{(h)})
\end{aligned}
$$
where $\boldsymbol{a}^{(in)}$ is a feature matrix of a training sample with dimensions $1 \times m$, and $W^{(h)}$ is a weight matrix with dimensions $m \times d$. Here, $m$ is the number of input features and $d$ is the number of units in the hidden layer. After matrix multiplication, we obtain a matrix of size $1 \times d$, which serves as the net input for computing the activation $\boldsymbol{a}^{(h)}$ ($\boldsymbol{a}^{(h)} \in \mathbb{R}^{1 \times d}$).

Furthermore, we generalize this to all $n$ samples in the training set:
$$
\begin{aligned}
\boldsymbol{Z}^{(h)} &= \boldsymbol{A}^{(in)}W^{(h)}
\end{aligned}
$$
where $\boldsymbol{A}^{(in)}$ is an $n \times m$ matrix. After multiplying it with the weight matrix, we obtain a matrix $\boldsymbol{Z}^{(h)}$ of size $n \times d$. Finally, we apply the activation function $\phi(.)$ to the net input matrix obtained in the previous step to produce a new activation matrix with the same dimensions:
$$
\begin{aligned}
\boldsymbol{A}^{(h)} &= \phi(\boldsymbol{Z}^{(h)})
\end{aligned}
$$

Similarly, for the output layer, we have the following matrix-form equations:
$$
\begin{aligned}
\boldsymbol{Z}^{(out)} &= \boldsymbol{A}^{(h)}W^{(out)}\\
\boldsymbol{A}^{(out)} &= \phi(\boldsymbol{Z}^{(out)}), \,\, \boldsymbol{A}^{(out)} \in \mathbb{R}^{n \times t}
\end{aligned}
$$
where $t$ is the number of output units.

### <font color='grey'>A brief note on backpropagation</font>
Although it was (re)discovered and published more than 30 years ago *(Learning representations by back-propagating errors, D. E. Rumelhart, G. E. Hinton, and R. J. Williams, Nature, 323:6088, pages 533–536, 1986)*, backpropagation remains one of the most widely used and effective algorithms for training artificial neural networks to this day.

In this section, we will briefly and intuitively review this algorithm. Essentially, backpropagation can be understood as an efficient way to compute the partial derivatives of a complex loss function in a multi-layer neural network. Our goal here is to use these partial derivatives to learn (update) the weights of the neural network model. One of the main challenges in parameterizing neural networks is that we often deal with a very large number of parameters with high dimensionality. Unlike the loss functions encountered in single-layer neural networks such as Adaline or Logistic Regression, the error surface of multi-layer neural networks is generally non-convex, non-smooth, and contains many local minima that must be navigated in order to find the global minimum of the loss function.

First, let us briefly recall the Chain Rule from calculus. The chain rule is an approach for solving the problem of finding the partial derivative of a complex function or a nested function, for example:
$$
\begin{aligned}
\frac{\partial}{\partial{x}}[h(f(x))] = \frac{\partial{h}}{\partial{f}} \cdot \frac{\partial f}{\partial x}
\end{aligned}
$$
Similarly, we can apply the chain rule to more complex functions. For example:

Suppose we have five different functions: $f(.), g(.), h(.), u(.)$, and $v(x)$. Let $F$ be the composition of these five functions:  
$F(x) = f(g(h(u(v(x)))))$.  
Applying the chain rule to compute the partial derivative of $F$ with respect to $x$, we obtain the following expression:
$$
\begin{aligned}
\frac{\partial F}{\partial x} = \frac{\partial}{\partial x}f(g(h(u(v(x))))) 
&= \frac{\partial f}{\partial g} \cdot \frac{\partial g}{\partial h} \cdot \frac{\partial h}{\partial u} \cdot \frac{\partial u}{\partial v} \cdot \frac{\partial v}{\partial x}
\end{aligned}
$$

### <font color='grey'>Training neural networks via backpropagation</font>
In this section, we will discuss some of the mathematics behind backpropagation to better understand how neural networks learn (update) the weights in the model.

First, in order to compute the loss and apply backpropagation, we need to sequentially perform feed-forward through the layers to obtain the output at the output layer. This process can be mathematically modeled as follows:
$$
\begin{aligned}
\boldsymbol{Z}^{(h)} &= \boldsymbol{A}^{(in)}W^{(h)}\,\,\text{(net input of the hidden layer) (1)}\\
\boldsymbol{A}^{(h)} &= \phi(\boldsymbol{Z}^{(h)}) \,\,\text{(activation of the hidden layer) (2)}\\
\boldsymbol{Z}^{(out)} &= \boldsymbol{A}^{(h)}W^{(out)}\,\,\text{(net input of the output layer) (3)}\\
\boldsymbol{A}^{(out)} &= \phi(\boldsymbol{Z}^{(out)}) \,\,\text{(activation of the output layer) (4)}\\
\end{aligned}
$$

This process is illustrated in the figure below:
![](https://raw.githubusercontent.com/HuangRihChang/machine_learning_basic/master/Deeplearning/images/ANN/FF3.png)

During backpropagation, we propagate the error backward from right to left (from the output layer to the input layer).  
Based on the chain rule, we have:
$$
\begin{aligned}
\frac{\partial J(\boldsymbol{W})}{\partial{\boldsymbol{W}^{(h)}}} &= 
\frac{\partial \boldsymbol{Z}^{(h)}}{\partial \boldsymbol{W}^{(h)}} 
\frac{\partial \boldsymbol{A}^{(h)}}{\partial \boldsymbol{Z}^{(h)}} 
\frac{\partial J(\boldsymbol{W})}{\partial \boldsymbol{A}^{(h)}}
\end{aligned}
$$
From equation (1), we can derive:
$$
\begin{aligned}
\frac{\partial \boldsymbol{Z}^{(h)}}{\partial \boldsymbol{W}^{(h)}} = \boldsymbol{A}^{(in)}
\end{aligned}
$$

Let us define $\frac{\partial \boldsymbol{A}^{(h)}}{\partial \boldsymbol{Z}^{(h)}} \frac{\partial J(\boldsymbol{W})}{\partial \boldsymbol{A}^{(h)}} = \delta^{(h)}$.  
Continuing the derivation, we obtain:
$$
\begin{aligned}
\delta^{(h)} = 
\frac{\partial \boldsymbol{A}^{(h)}}{\partial \boldsymbol{Z}^{(h)}} 
\frac{\partial J(\boldsymbol{W})}{\partial \boldsymbol{A}^{(h)}} 
= \frac{\partial \boldsymbol{A}^{(h)}}{\partial \boldsymbol{Z}^{(h)}} 
\frac{\partial \boldsymbol{Z}^{(out)}}{\partial \boldsymbol{A}^{(h)}} 
\frac{\partial J(\boldsymbol{W})}{\partial \boldsymbol{Z}^{(out)}}
\end{aligned}
$$

We arrive at the following formula:
$$
\begin{aligned}
\delta^{(h)} &= \delta^{(out)}(\boldsymbol{W}^{(out)})^T \odot 
\frac{\partial \phi(z^{(h)})}{\partial z^{(h)}}
\end{aligned}
$$

Here, $\frac{\partial \phi(z^{(h)})}{\partial z^{(h)}}$ is the derivative of the activation function, which we will compute for some basic activation functions in the next section. For simplicity, we assume that the activation function used is the sigmoid function, whose derivative is given by:
$$
\begin{aligned}
\frac{\partial \phi(z)}{\partial z} = \phi'(z) = \phi(z)\odot(1-\phi(z))
\end{aligned}
$$

The term $\delta^{(out)}$ can be computed as follows (proof available at this [link](https://math.stackexchange.com/questions/945871/derivative-of-softmax-loss-function)):
$$
\begin{aligned}
\delta^{(out)} = a^{(out)} - y
\end{aligned}
$$
where $y$ is the matrix of true labels.

The operator $\odot$ in the above equations denotes element-wise multiplication.

Thus, $\delta^{(h)}$ can be written as:
$$
\begin{aligned}
\delta^{(h)} &= \delta^{(out)}(\boldsymbol{W}^{(out)})^T \odot 
(a^{(h)} \odot (1-a^{(h)}))
\end{aligned}
$$

We can interpret the above computation as:  
$(n, h) = (n, c)\cdot(c, h) \odot (n, h)$,  
where $n$ is the number of training samples, $c$ is the number of output units, and $h$ is the number of hidden activation units.

To facilitate implementation and experimentation, we vectorize the gradients as follows:
$$
\begin{aligned}
\Delta^{(h)} = (\boldsymbol{A}^{(in)})^T \delta^{(h)}\\
\Delta^{(out)} = (\boldsymbol{A}^{(h)})^T \delta^{(out)}
\end{aligned}
$$

Finally, after computing the gradients, we can update the weights of the model at each layer $l$ by taking a step in the opposite direction of the gradient:
$$
\begin{aligned}
\boldsymbol{W}^{(l)} := \boldsymbol{W}^{(l)} - \eta \Delta^{(l)}
\end{aligned}
$$

#### Common activation functions

**Sigmoid**
$$
\begin{aligned}
\sigma(s) = \frac{1}{1+e^{-s}}
\end{aligned}
$$

**Swish/SiLU**
$$
\begin{aligned}
\text{swish}(s) = s \cdot \sigma(s) = \frac{s}{1+e^{-s}}
\end{aligned}
$$

**Tanh**
$$
\begin{aligned}
\text{tanh}(s) = 2\sigma(2s)-1 = \frac{e^s-e^{-s}}{e^s+e^{-s}}
\end{aligned}
$$

**ReLU**
$$
\begin{aligned}
 \text{relu}(s) = \text{max}(0,s)
\end{aligned}
$$

**Note**: The above functions are applied to a real number $(s \in \mathbb{R})$. When the input is a vector, these functions are applied element-wise.

------------------------------------
**Softmax**
$$
\begin{aligned}
\sigma(\mathbf{z})_{i} = \frac{e^{\beta z_{i}}}{\sum_{j=1}^{K} e^{\beta z_{j}}}
\quad \text{for } i = 1, \dotsc, K
\end{aligned}
$$
